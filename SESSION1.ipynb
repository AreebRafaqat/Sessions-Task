# AI Use-Case Canvas
**Scenario:** IT Helpdesk — Ticket Triage & First-Response
**Use case:** AI-assisted support ticket sorting and reply drafting

## 1. Process & Pain Point
When staff have an IT problem (password reset, software error, hardware issue, network problem), they send an email or fill a form to the IT helpdesk. A small IT team reads each ticket, works out how urgent and what type it is, then replies or assigns it to the right person.

**Pain point:** Tickets pile up fast. Many are simple and repeat (like "forgot my password"), but they still take time to read, sort, and reply to one by one. Urgent tickets (like "whole system is down") can get buried under a pile of small ones, causing delays.

## 2. Data Needed & Data Risks
**Data needed:** Past ticket text and replies, ticket categories (hardware/software/network/access), priority labels, and response times from before.

**Data risks:**
- Tickets may contain usernames, passwords, or personal staff details — must be handled securely
- Risk of a data leak if ticket data is sent to an AI tool outside company approval
- Old tickets may have out-of-date info (staff who left, old systems) that could confuse the AI
- Must follow company IT security and data protection policy

## 3. Best First Approach: Predictive ML or GenAI?
**A mix, but start with predictive ML for sorting, GenAI for replies.**

- Predictive ML = AI that predicts a category or urgency level from past data (like sorting tickets into "urgent" vs "normal")
- GenAI = AI that writes text (like drafting a reply or instructions)

Since the main pain point is fast, accurate sorting, predictive ML is the first step (auto-tag ticket type and urgency). GenAI comes second, to draft a first-response reply for the common, simple tickets (like password resets), which a person still checks before sending.

## 4. How Colleagues Would Use the Output
IT staff see tickets already sorted by type and urgency when they open the queue, so urgent issues stand out immediately. For common simple tickets, they get an AI-drafted reply they can quickly check and send, instead of typing it from scratch each time.

## 5. How to Measure Success
- Faster average response time, especially for urgent tickets
- Fewer tickets sitting unanswered for a long time
- Staff feedback: is the sorting accurate? Are drafted replies useful?
- Fewer tickets wrongly marked as low priority when they were actually urgent

## 6. Governance Risk & Mitigation
**Risk:** The AI could wrongly mark a serious issue (like a security breach or system outage) as low priority, delaying a proper fix.

**Mitigation:** Keep a person in charge of final priority checks, especially for anything flagged as "uncertain" by the AI. Set clear rules so certain keywords (like "hacked," "breach," "down for everyone") always get flagged for a human to check immediately, no matter what the AI predicts.
